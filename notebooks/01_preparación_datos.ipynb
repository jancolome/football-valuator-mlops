# Football Valuator MLOps — Preparación de datos

Notebook de exploración y preparación del dataset para el modelo de predicción del valor de mercado de futbolistas.

**Fuente:** Transfermarkt (Kaggle)

**Output esperado:** un CSV con una fila por cada valoración de jugador, con stats de la ventana entre valoraciones.

```python
import pandas as pd

# Configuración para que pandas muestre todas las columnas
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)

# Rutas a los CSVs
RUTA_PLAYERS = '../data/raw/players.csv'
RUTA_VALUATIONS = '../data/raw/player_valuations.csv'
RUTA_APPEARANCES = '../data/raw/appearances.csv'

print("✓ Configuración cargada")
```

```python
print("📂 Cargando players.csv...")
players = pd.read_csv(RUTA_PLAYERS)
print(f"   ✓ {len(players):,} jugadores cargados")
print(f"   ✓ {len(players.columns)} columnas")

# Ver las primeras 5 filas
players.head()
```

```python
print("📂 Cargando appearances.csv...")
appearances = pd.read_csv(RUTA_APPEARANCES)
print(f"   ✓ {len(appearances):,} apariciones cargadas")
print(f"   ✓ {len(appearances.columns)} columnas")

appearances.head()
```

```python
# Convertir las columnas de fecha de texto (object) a fecha (datetime)
players['date_of_birth'] = pd.to_datetime(players['date_of_birth'], errors='coerce')
valuations['date'] = pd.to_datetime(valuations['date'], errors='coerce')
appearances['date'] = pd.to_datetime(appearances['date'], errors='coerce')

print("✓ Fechas convertidas")
print(f"\nplayers.date_of_birth: {players['date_of_birth'].dtype}")
print(f"valuations.date: {valuations['date'].dtype}")
print(f"appearances.date: {appearances['date'].dtype}")
```

```python
# Filtrar solo valoraciones desde 2012 (mejor calidad de datos)
fecha_corte = pd.Timestamp('2012-01-01')

valuations_filtradas = valuations[valuations['date'] >= fecha_corte].copy()

print(f"Valuations originales: {len(valuations):,}")
print(f"Valuations desde 2012: {len(valuations_filtradas):,}")
print(f"Reducción: {(1 - len(valuations_filtradas)/len(valuations))*100:.1f}%")
```

```python
# Eliminar valoraciones con valor 0 o nulo (no aportan)
valuations_filtradas = valuations_filtradas[valuations_filtradas['market_value_in_eur'] > 0].copy()

print(f"Valuations después de quitar valor 0: {len(valuations_filtradas):,}")
print("\nEstadísticas del valor:")
pd.set_option('display.float_format', lambda x: f'{x:,.0f}')
print(valuations_filtradas['market_value_in_eur'].describe())
```

```python
# ============================================================
# QUITAR DATOS DE 2026 (INCOMPLETOS)
# ============================================================

# Filtrar valuations: solo hasta 2025
antes_val = len(valuations_filtradas)
valuations_filtradas = valuations_filtradas[valuations_filtradas['date'].dt.year <= 2025].copy()
print(f"Valuations: {antes_val:,} → {len(valuations_filtradas):,} (eliminadas {antes_val - len(valuations_filtradas):,})")

# Filtrar appearances: solo hasta 2025
antes_app = len(appearances)
appearances = appearances[appearances['date'].dt.year <= 2025].copy()
print(f"Appearances: {antes_app:,} → {len(appearances):,} (eliminadas {antes_app - len(appearances):,})")
```

```python
# ============================================================
# ÍNDICE DE INFLACIÓN DEL MERCADO DEL FÚTBOL
# ============================================================
# Calculamos para cada año el valor mediano del top 100 jugadores
# Eso nos da una "referencia" del valor del mercado en ese año

# 1. Añadir el año a cada valoración
valuations_filtradas['año'] = valuations_filtradas['date'].dt.year

# 2. Para cada (jugador, año), coger su valoración MÁXIMA en ese año
valor_max_anual = (valuations_filtradas
    .groupby(['player_id', 'año'])['market_value_in_eur']
    .max()
    .reset_index()
)

# 3. Para cada año, coger el TOP 100 jugadores y calcular el valor mediano
top100_por_año = (valor_max_anual
    .sort_values(['año', 'market_value_in_eur'], ascending=[True, False])
    .groupby('año')
    .head(100)
    .groupby('año')['market_value_in_eur']
    .median()
    .reset_index()
    .rename(columns={'market_value_in_eur': 'valor_mediano_top100'})
)

print("📈 Valor mediano del TOP 100 por año:")
print(top100_por_año.to_string(index=False))
```

```python
# Para cada valoración, calcular la fecha de la valoración ANTERIOR del mismo jugador
valuations_filtradas['fecha_anterior'] = valuations_filtradas.groupby('player_id')['date'].shift(1)

# Para la primera valoración de cada jugador, no hay anterior
# Usamos como inicio la fecha de la valoración menos 1 año (por defecto)
# Esto se gestionará después

print("✓ Columna 'fecha_anterior' añadida")
print(f"\nEjemplo (player_id=10, Klose):")
valuations_filtradas[valuations_filtradas['player_id'] == 10][['player_id', 'date', 'fecha_anterior', 'market_value_in_eur']].head(10)
```

```python
# Para las primeras valoraciones (sin anterior), usar fecha_actual - 2 años
import pandas as pd

mask_primera = valuations_filtradas['fecha_anterior'].isna()
valuations_filtradas.loc[mask_primera, 'fecha_anterior'] = (
    valuations_filtradas.loc[mask_primera, 'date'] - pd.DateOffset(years=2)
)

print(f"Primeras valoraciones rellenadas: {mask_primera.sum():,}")
print(f"\nVerificación (Klose primera valoración):")
valuations_filtradas[valuations_filtradas['player_id'] == 10].head(3)
```

```python
# ============================================================
# CALCULAR COEFICIENTE DE INFLACIÓN (BASE: 2025)
# ============================================================
# Coef 2025 = 1.0 (referencia)
# Coef 2012 = 75M / 30M = 2.5 → un valor de 2012 hay que multiplicarlo por 2.5

VALOR_BASE = top100_por_año[top100_por_año['año'] == 2025]['valor_mediano_top100'].iloc[0]
top100_por_año['coef_inflacion'] = VALOR_BASE / top100_por_año['valor_mediano_top100']

# Mostrar con decimales para verificar
pd.set_option('display.float_format', lambda x: f'{x:,.2f}')

print(f"📊 Año base: 2025 ({VALOR_BASE:,.0f} €)")
print(f"\nCoeficientes de inflación (base 2025 = 1.00):")
print(top100_por_año.to_string(index=False))

# Volvemos al formato sin decimales
pd.set_option('display.float_format', lambda x: f'{x:,.0f}')
```

```python
# ============================================================
# APLICAR EL COEFICIENTE DE INFLACIÓN AL VALOR
# ============================================================

# Si la columna ya existe (re-ejecución), la quitamos primero
if 'coef_inflacion' in valuations_filtradas.columns:
    valuations_filtradas = valuations_filtradas.drop(columns=['coef_inflacion'])
if 'valor_ajustado_2025' in valuations_filtradas.columns:
    valuations_filtradas = valuations_filtradas.drop(columns=['valor_ajustado_2025'])

# Unir el coeficiente a cada valoración según su año
valuations_filtradas = valuations_filtradas.merge(
    top100_por_año[['año', 'coef_inflacion']],
    on='año',
    how='left'
)

# Calcular el valor ajustado (en euros 2025 equivalentes)
valuations_filtradas['valor_ajustado_2025'] = (
    valuations_filtradas['market_value_in_eur'] * valuations_filtradas['coef_inflacion']
).round(0)

print("✓ Valor ajustado a inflación calculado")
print(f"\nMuestra (Klose):")
valuations_filtradas[valuations_filtradas['player_id'] == 10][
    ['date', 'año', 'market_value_in_eur', 'coef_inflacion', 'valor_ajustado_2025']
].head(10)
```

```python
# ============================================================
# CALCULAR STATS DE LA VENTANA ENTRE VALORACIONES
# ============================================================
# Para cada valoración: sumar goles, asistencias, partidos, minutos
# de todos los partidos que ocurrieron entre [fecha_anterior, date]

print("⚙️  Calculando stats de la ventana... (puede tardar 1-3 min)")

# 1. Preparar valuations con un id único por valoración
val_simple = valuations_filtradas[['player_id', 'date', 'fecha_anterior']].copy()
val_simple = val_simple.reset_index(drop=False).rename(columns={'index': 'valuation_idx'})

# 2. Asignar cada partido a la valoración correspondiente usando merge_asof
#    (busca para cada partido la primera valoración POSTERIOR del mismo jugador)
print("   • Asignando partidos a ventanas...")

apps_sorted = appearances.sort_values('date').rename(columns={'date': 'fecha_partido'})
val_sorted = val_simple.sort_values('date')

matched = pd.merge_asof(
    apps_sorted,
    val_sorted,
    by='player_id',
    left_on='fecha_partido',
    right_on='date',
    direction='forward'
)

# 3. Filtrar partidos que caen DENTRO de la ventana válida
matched = matched[
    (matched['fecha_partido'] >= matched['fecha_anterior']) &
    (matched['fecha_partido'] <= matched['date'])
].copy()

print(f"   • Partidos asignados a ventanas: {len(matched):,}")

# 4. Agrupar por valoración y calcular stats
print("   • Agrupando stats por ventana...")

stats_ventana = matched.groupby('valuation_idx').agg(
    goles_ventana=('goals', 'sum'),
    asistencias_ventana=('assists', 'sum'),
    minutos_ventana=('minutes_played', 'sum'),
    amarillas_ventana=('yellow_cards', 'sum'),
    rojas_ventana=('red_cards', 'sum'),
    partidos_ventana=('appearance_id', 'count'),
    liga_predominante=('competition_id', lambda x: x.mode().iloc[0] if not x.mode().empty else None)
)

print(f"✓ Stats calculadas para {len(stats_ventana):,} valoraciones")
print(f"\nMuestra:")
stats_ventana.head()
```

```python
# Ver una muestra aleatoria con jugadores conocidos
muestra = stats_ventana.sample(20, random_state=42)
print(muestra.to_string())

print(f"\n\nEstadísticas globales:")
print(stats_ventana.describe())
```

```python
# ============================================================
# UNIR LAS STATS DE LA VENTANA CON LAS VALORACIONES
# ============================================================

# Reset index para tener valuation_idx como columna
valuations_filtradas = valuations_filtradas.reset_index(drop=True)
valuations_filtradas['valuation_idx'] = valuations_filtradas.index

# Unir con las stats calculadas
valuations_con_stats = valuations_filtradas.merge(
    stats_ventana,
    on='valuation_idx',
    how='left'
)

# Rellenar con 0 los que no tienen stats (no jugaron en la ventana)
columnas_stats = ['goles_ventana', 'asistencias_ventana', 'minutos_ventana',
                  'amarillas_ventana', 'rojas_ventana', 'partidos_ventana']
valuations_con_stats[columnas_stats] = valuations_con_stats[columnas_stats].fillna(0).astype(int)

# Para liga_predominante, los nulos los marcamos como "DESCONOCIDA"
valuations_con_stats['liga_predominante'] = valuations_con_stats['liga_predominante'].fillna('DESCONOCIDA')

print(f"✓ Stats unidas a valuations")
print(f"\nTotal de filas: {len(valuations_con_stats):,}")
print(f"\nDistribución partidos_ventana:")
print(valuations_con_stats['partidos_ventana'].describe())
print(f"\nValoraciones sin partidos en la ventana: {(valuations_con_stats['partidos_ventana']==0).sum():,} ({(valuations_con_stats['partidos_ventana']==0).mean()*100:.1f}%)")
```

```python
# ============================================================
# FILTRAR VALORACIONES SIN PARTIDOS EN LA VENTANA
# ============================================================

antes = len(valuations_con_stats)
valuations_con_stats = valuations_con_stats[valuations_con_stats['partidos_ventana'] > 0].copy()
despues = len(valuations_con_stats)

print(f"Valoraciones con stats: {antes:,} → {despues:,}")
print(f"Eliminadas (sin partidos): {antes - despues:,} ({(antes-despues)/antes*100:.1f}%)")
print(f"\nDistribución partidos_ventana (después del filtro):")
print(valuations_con_stats['partidos_ventana'].describe())
```

```python
# ============================================================
# UNIR DATOS DEL JUGADOR
# ============================================================

# Columnas útiles de players
columnas_player = [
    'player_id',
    'name',
    'date_of_birth',
    'position',
    'sub_position',
    'foot',
    'height_in_cm',
    'country_of_citizenship',
    'international_caps'
]

# Hacemos el merge
dataset = valuations_con_stats.merge(
    players[columnas_player],
    on='player_id',
    how='left'
)

# Eliminar filas sin datos de jugador (raras pero pueden existir)
antes = len(dataset)
dataset = dataset.dropna(subset=['date_of_birth', 'position'])
despues = len(dataset)

print(f"✓ Datos del jugador unidos")
print(f"   Filas: {antes:,} → {despues:,} (eliminadas {antes-despues:,} sin datos básicos)")
print(f"\nColumnas del dataset:")
print(dataset.columns.tolist())
```

```python
# ============================================================
# CREAR FEATURES FINALES
# ============================================================

# 1. Quitar columnas que no usamos en el modelo
columnas_a_quitar = [
    'current_club_id',  # no fiable (es el actual)
    'player_club_domestic_competition_id',  # no fiable (es el actual)
    'valuation_idx',  # solo era para el merge
]
dataset = dataset.drop(columns=[c for c in columnas_a_quitar if c in dataset.columns])

# 2. Edad en el momento de la valoración
dataset['edad'] = ((dataset['date'] - dataset['date_of_birth']).dt.days / 365.25).round(2)

# 3. Edad al cuadrado (efecto no-lineal: explosión joven, declive viejo)
dataset['edad_cuadrado'] = (dataset['edad'] ** 2).round(2)

# 4. Es internacional (sí si tiene caps registrados)
dataset['es_internacional'] = dataset['international_caps'].fillna(0) > 0

# 5. Goles por minuto
dataset['goles_por_minuto'] = (dataset['goles_ventana'] / dataset['minutos_ventana'].replace(0, 1)).round(5)

# 6. Asistencias por minuto
dataset['asistencias_por_minuto'] = (dataset['asistencias_ventana'] / dataset['minutos_ventana'].replace(0, 1)).round(5)

# 7. Ratio de minutos: minutos jugados / minutos posibles (90 × partidos)
dataset['ratio_minutos'] = (dataset['minutos_ventana'] / (dataset['partidos_ventana'] * 90).replace(0, 1)).round(3)

# 8. TIER DE LIGA
LIGAS_TIER_1 = ['GB1', 'ES1', 'L1', 'IT1', 'FR1']
LIGAS_TIER_2 = ['NL1', 'PO1', 'BE1', 'TR1', 'RU1']
LIGAS_UEFA = ['CL', 'EL', 'ECL', 'CLQ', 'ELQ', 'ECLQ']

def asignar_tier(liga):
    if liga in LIGAS_TIER_1:
        return 1
    elif liga in LIGAS_TIER_2:
        return 2
    elif liga in LIGAS_UEFA:
        return 99  # UEFA = especial
    else:
        return 3

dataset['tier_liga'] = dataset['liga_predominante'].apply(asignar_tier)

print(f"✓ Features creadas")
print(f"\nMuestra (Klose):")
muestra = dataset[dataset['player_id'] == 10][[
    'date', 'edad', 'goles_ventana', 'minutos_ventana',
    'goles_por_minuto', 'ratio_minutos', 'tier_liga',
    'market_value_in_eur', 'valor_ajustado_2025'
]].head(5)
print(muestra.to_string())

print(f"\nDistribución tier_liga:")
print(dataset['tier_liga'].value_counts().sort_index())
```

```python
# ============================================================
# GUARDAR DATASET FINAL
# ============================================================

# Columnas finales para el modelo
columnas_modelo = [
    # Identificadores (para depurar, no son features)
    'player_id', 'name', 'date',
    
    # Target (lo que queremos predecir)
    'valor_ajustado_2025',  # ya ajustado a inflación 2025
    'market_value_in_eur',  # original sin ajustar (por si lo queremos comparar)
    
    # Features del jugador
    'edad', 'edad_cuadrado', 'position', 'sub_position', 
    'foot', 'height_in_cm', 'country_of_citizenship', 'es_internacional',
    
    # Features de la ventana
    'goles_ventana', 'asistencias_ventana', 'partidos_ventana', 
    'minutos_ventana', 'amarillas_ventana', 'rojas_ventana',
    'goles_por_minuto', 'asistencias_por_minuto', 'ratio_minutos',
    'tier_liga'
]

dataset_final = dataset[columnas_modelo].copy()

# Guardar
ruta_salida = 'data/processed/dataset_modelo.csv'
dataset_final.to_csv(ruta_salida, index=False)

print(f"✓ Dataset guardado en: {ruta_salida}")
print(f"   Filas: {len(dataset_final):,}")
print(f"   Columnas: {len(dataset_final.columns)}")
print(f"\nPrimeras 5 filas:")
dataset_final.head()
```
